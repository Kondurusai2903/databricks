#### Process Customers Data

- Ingest the Data into data lake house - bronze_customers
- Perform data quality checks and transform the data as required - silver_customer_clean
- Apply changes to the customers data - silver_customers

```python
%sql
use catalog circuitbox;
use schema landing;
```

```python
%sql
CREATE OR REFRESH STREAMING TABLE bronze_customers
COMMENT 'The is a bronze streaming customers Table'
TBLPROPERTIES ('quality'='bronze')
AS
SELECT *, _metadata.file_path as file_path ,current_timestamp as ingestion_timestamp from
cloud_files(
    '/Volumes/circuitbox/landing/operational_data/customers',
    'json'
    'cloudFiles.inferColumnTypes'='true'
)
```
