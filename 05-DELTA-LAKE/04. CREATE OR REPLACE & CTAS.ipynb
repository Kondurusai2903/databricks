##### Behaviour of DROP and CREATE Statements

```python
%sql
DROP TABLE IF EXISTS gtesting.delta_lake.companies;

CREATE TABLE IF NOT EXISTS gtesting.delta_lake.companies (
    company_id BIGINT NOT NULL GENERATED ALWAYS as IDENTITY (START WITH 1 INCREMENT BY 1),
    company_name STRING,
    founded_date DATE COMMENT 'This Date is Very Special for the Company',
    foundation_year INT GENERATED ALWAYS as (year(founded_date)) COMMENT 'This is the Year of Foundation',
    country STRING
)
COMMENT 'The table contains information about some of the successfull companies in the world'
TBLPROPERTIES ('sensitive'='true','delta.enableDeletionVectors'='false');



INSERT INTO gtesting.delta_lake.companies (company_name, founded_date, country) VALUES ('Microsoft', '1975-04-04', 'USA'), ('Google', '1998-09-04', 'USA'), ('Amazon', '1994-05-17', 'USA'), ('Alibaba', '1999-07-01', 'China'), ('Tencent', '1998-12-17', 'China'), ('SpaceX', '2002-06-30', 'USA'), ('Uber', '2009-03-31', 'USA'), ('Lyft', '2012-03-03', 'USA'), ('Airbnb', '2008-08-01', 'USA'), ('PayPal', '1998-01-01', 'USA'), ('Square', '2009-01-01', 'USA');
```

##### Behaviour of CREATE OR REPLACE STATEMENTS

```python
%sql
DROP TABLE IF EXISTS gtesting.delta_lake.companies;
```

```python
%sql
CREATE OR REPLACE TABLE gtesting.delta_lake.companies (
    company_id BIGINT NOT NULL GENERATED ALWAYS as IDENTITY (START WITH 1 INCREMENT BY 1),
    company_name STRING,
    founded_date DATE COMMENT 'This Date is Very Special for the Company',
    foundation_year INT GENERATED ALWAYS as (year(founded_date)) COMMENT 'This is the Year of Foundation',
    country STRING
)
COMMENT 'The table contains information about some of the successfull companies in the world'
TBLPROPERTIES ('sensitive'='true','delta.enableDeletionVectors'='false');



INSERT INTO gtesting.delta_lake.companies (company_name, founded_date, country) VALUES ('Microsoft', '1975-04-04', 'USA'), ('Google', '1998-09-04', 'USA'), ('Amazon', '1994-05-17', 'USA'), ('Alibaba', '1999-07-01', 'China'), ('Tencent', '1998-12-17', 'China'), ('SpaceX', '2002-06-30', 'USA'), ('Uber', '2009-03-31', 'USA'), ('Lyft', '2012-03-03', 'USA'), ('Airbnb', '2008-08-01', 'USA'), ('PayPal', '1998-01-01', 'USA'), ('Square', '2009-01-01', 'USA');
```

```python
%sql
desc history gtesting.delta_lake.companies;
```

##### CTAs Statement

```python
%sql
CREATE TABLE gtesting.delta_lake.china_companies
as 
select * from gtesting.delta_lake.companies where country = 'China';
```

```python
%sql
select * from gtesting.delta_lake.china_companies;
```

```python
%sql
desc gtesting.delta_lake.china_companies;
```
