### INSERT OVERWRITE

- Replace all the data in a table
- Replace all the data from a specific partition
- How to handle Schema Changes



> INSERT OVERWRITE - overwrite the existing data in a table or a specific partition with new data 

> INSERT INTO - Appends into


##### Replace all the data in a table

```python
%sql

DROP TABLE IF EXISTS gtesting.delta_lake.gold_companies;

create table gtesting.delta_lake.gold_companies
(
    company_name STRING,
    founded_data DATE,
    country STRING
);


INSERT INTO gtesting.delta_lake.gold_companies
VALUES ('Microsoft', '1975-04-04', 'USA'), ('Google', '1998-09-04', 'USA'), ('Alibaba', '1999-07-01', 'China');
```

```python
%sql
select * from gtesting.delta_lake.gold_companies;
```

```python
%sql

DROP TABLE IF EXISTS gtesting.delta_lake.bronze_companies;

create table gtesting.delta_lake.bronze_companies
(
    company_name STRING,
    founded_data DATE,
    country STRING
);


INSERT INTO gtesting.delta_lake.bronze_companies
VALUES ('Microsoft', '1975-04-04', 'USA'), ('Google', '1998-09-04', 'USA'), ('Alibaba', '1999-07-01', 'China'),('Oracle', '1975-04-04', 'USA'),('Databricks', '1975-04-04', 'USA');
```

```python
%sql
INSERT OVERWRITE gtesting.delta_lake.gold_companies
SELECT * FROM gtesting.delta_lake.bronze_companies;
    

select * from gtesting.delta_lake.gold_companies;
```

##### Replace all the data from a specific partition

```python
%sql
DROP TABLE IF EXISTS gtesting.delta_lake.gold_companies_partition;

create table gtesting.delta_lake.gold_companies_partition
(
    company_name STRING,
    founded_data DATE,
    country STRING
) PARTITIONED BY (country);


INSERT INTO gtesting.delta_lake.gold_companies_partition
VALUES ('Microsoft', '1975-04-04', 'USA'), ('Google', '1998-09-04', 'USA'), ('Alibaba', '1999-07-01', 'China');
```

```python
%sql
select * from gtesting.delta_lake.gold_companies_partition;
```

```python
%sql
DESC EXTENDED gtesting.delta_lake.gold_companies_partition;
```

```python
%sql
DROP TABLE IF EXISTS gtesting.delta_lake.bronze_companies_usa;

create table gtesting.delta_lake.bronze_companies_usa
(
    company_name STRING,
    founded_data DATE,
    country STRING
) PARTITIONED BY (country);


INSERT INTO gtesting.delta_lake.bronze_companies_usa
VALUES ('Microsoft', '1975-04-04', 'USA'), ('Google', '1998-09-04', 'USA');
```

```python
%sql
INSERT OVERWRITE gtesting.delta_lake.gold_companies_partition
PARTITION (country='USA')
SELECT company_name, founded_data FROM gtesting.delta_lake.bronze_companies_usa;
```

```python
%sql
select * from gtesting.delta_lake.gold_companies_partition;
```
