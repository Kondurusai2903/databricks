
### Create Table - Table & Column Properties

Demonstrate adding Table and Column Properties to the CREATE TABLE statement

#### 1. Table Properties
> 1.1 COMMENT - allows you to document the purpose of  table

> 1.2 TBLPROPERTIES - used to specify table level metadata or configuration settings

```python
%sql

DROP TABLE IF EXISTS gtesting.delta_lake.companies;

CREATE SCHEMA IF NOT EXISTS gtesting.delta_lake;

CREATE TABLE IF NOT EXISTS gtesting.delta_lake.companies (
    company_name STRING,
    founded_date DATE,
    country STRING
)
COMMENT 'The table contains information about some of the successfull companies in the world'
TBLPROPERTIES ('sensitive'='true','delta.enableDeletionVectors'='false');

```

```python
%sql
DESC EXTENDED gtesting.delta_lake.companies
```

#### 2. Column Properties

> 2.1 NOT NULL CONSTRAINTS - enforces data integrity and quality by ensuring that a specific column cannot contains NULL Values

> 2.2 COMMENT - documents the purpose of context of individual columns in a table 

```python
%sql
DROP TABLE IF EXISTS gtesting.delta_lake.companies;

CREATE TABLE IF NOT EXISTS gtesting.delta_lake.companies (
    company_name STRING,
    founded_date DATE COMMENT 'This Date is Very Special for the Company',
    country STRING
)
COMMENT 'The table contains information about some of the successfull companies in the world'
TBLPROPERTIES ('sensitive'='true','delta.enableDeletionVectors'='false');

```

```python
%sql
DESC EXTENDED gtesting.delta_lake.companies
```

#### Column Properties

> 2.3 Generated Columns - derived or computed Columns, whos values are computed at the time of inserting a new records

      > 2.3.1 Generated Identity Columns - Used to generate an identity for example primary key value 

      > 2.3.2 Generated Compute Columns - automatically calcutates and store derived values based on other columns in the same


#### 2.3.1 Generated Identity Columns 
> Generated { ALWAYS | BY DEFAULT } AS IDENTITY  [ ( [ START WITH start ] [INCREMENT BY step])]

```python
%sql
DROP TABLE IF EXISTS gtesting.delta_lake.companies;

CREATE TABLE IF NOT EXISTS gtesting.delta_lake.companies (
    company_id BIGINT NOT NULL GENERATED ALWAYS as IDENTITY (START WITH 1 INCREMENT BY 1),
    company_name STRING,
    founded_date DATE COMMENT 'This Date is Very Special for the Company',
    country STRING
)
COMMENT 'The table contains information about some of the successfull companies in the world'
TBLPROPERTIES ('sensitive'='true','delta.enableDeletionVectors'='false');
```

#### 2.3.2 Generated Computed Columns 

> GENERATED ALWAYS AS (expr)

> expr 
may be composed of literals, column identifiers within the table, and determistic , built in SQL-Functions or operators expect:


- Aggregate Functions
- Analytic Window Functions
- Ranking Window Functions
- Table Valued Generator Functions

Also expr must not contain any subquery




```python
%sql
DROP TABLE IF EXISTS gtesting.delta_lake.companies;

CREATE TABLE IF NOT EXISTS gtesting.delta_lake.companies (
    company_id BIGINT NOT NULL GENERATED ALWAYS as IDENTITY (START WITH 1 INCREMENT BY 1),
    company_name STRING,
    founded_date DATE COMMENT 'This Date is Very Special for the Company',
    founded_year INT GENERATED ALWAYS as (year(founded_date)),
    country STRING
)
COMMENT 'The table contains information about some of the successfull companies in the world'
TBLPROPERTIES ('sensitive'='true','delta.enableDeletionVectors'='false');
```

```python
%sql
INSERT INTO gtesting.delta_lake.companies (company_name, founded_date, country) VALUES ('Microsoft', '1975-04-04', 'USA'), ('Google', '1998-09-04', 'USA'), ('Amazon', '1994-05-17', 'USA'), ('Alibaba', '1999-07-01', 'China'), ('Tencent', '1998-12-17', 'China'), ('SpaceX', '2002-06-30', 'USA'), ('Uber', '2009-03-31', 'USA'), ('Lyft', '2012-03-03', 'USA'), ('Airbnb', '2008-08-01', 'USA'), ('PayPal', '1998-01-01', 'USA'), ('Square', '2009-01-01', 'USA');
```

```python
%sql
select * from gtesting.delta_lake.companies;
```
